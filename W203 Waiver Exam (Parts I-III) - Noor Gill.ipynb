**1:** This is a case of multiple testing since there is a set of hypotheses being tested simultaneously - whether cat owners sleep longer than people without pets and whether dog owners sleep longer than people without pets. However, the sample sizes of the groups are not equal since there are 60 cat owners, 30 dog owners, and 100 people without pets in the dataset used for the nominal t-test. This results in an inflated error rate since the sample sizes in all 3 conditions are unequal, which increases the Type 1 error rate of the t-test relative to the nominal alpha level. A high Type 1 Error Rate is often observed if there are unequal variances and unequal sample sizes among the groups, as in this example.

On the other hand, if the sample sizes in all 3 conditions were equal, the t-test would be robust against unequal variances and the error rate would not be inflated. Another way to reduce the Type 1 Error would be to perform a Welch's t-test instead since there is unequal variance in the dataset.

**2:** When homoscedasticity of errors is violated, there is heteroscedasticity of errors present, which makes it so the standard deviation of forecasted errors is hard to determine and confidence intervals can be too narrow or wide as a result. Moreover, when homoscedasticity is violated, in estimating coefficients, there may be too much weight placed on a portion of the data where the error variance is greater relative to other segments of the data.  

A strategy to check if the homoskedasticity of errors assumption holds in the Classical Linear Model is viewing a plot of the residuals versus the predicted values. If residuals grow larger as a function of the predicted value, the data is heteroscedastic. For the homoscedasticity assumption to hold, there should be no clear patterns in the distribution. To verify the assumption is met, ideally, the residuals should be equally spread around the line, y = 0.

Another strategy is generating a plot of residuals versus independent variables. Although the errors may be larger for forecasts that are associated with predictions or independent variable values that are extreme in both directions due to a lack of precision in the estimates of the coefficients, this effect should not be too large. In other words, if the errors do not get systematically greater by a significant amount in a particular direction, the homoskedasticity assumption holds and it does not otherwise. 

If I detected a violation of homoskedasticity in the course of conducting a linear regression, I would respond based on the context and nature of the dataset. For example, if the residual versus predicted values plot displays that the size of the errors is proportional to the size of the predictions such that errors are consistent in terms of percent but not necessarily in absolute term or if the dependent variable is strictly positive, I would apply a log transformation to the dependent variable. Since a violation of the linearity assumption or the independence assumption in linear regression can also lead to heteroscedasticity, I would ensure those assumptions are satisfied as well, which may tackle the issue of ensuring homoscedasticity as well.


**3:** In the context of Classical Linear Regression, serial correlation of errors occurs when error terms from different time periods or cross-section observations are correlated so the error term is serially correlated. This violates the assumption that observations of the error term are uncorrelated with each other. The Durbin-Watson statistic tests for serial correlation, on a scale from 0 to 4.

Pure serial correlation does not result in biased regression coefficient estimates. Although serial correlation does not affect the unbiasedness or consistency of OLS estimators, it affects their efficiency. When there is a positive serial correlation, the OLS estimates of the standard errors are smaller than the true values for the standard errors, which allows the parameter estimates to be concluded as more precise than they actually are, hence there is an inclination to reject the null hypothesis even though it should not be rejected. It causes the estimated variances of the regression coefficients to be biased so the t-statistic appears more significant than it actually is, which makes hypothesis testing unreliable. Another problem that occurs is that serial correlation makes it so that the OLS is no longer a minimum variance estimator.

**4:** In the context of Classical Linear Regression, an omitted variable is any variable not included as an independent variable in the regression, but one that may influence the dependent variable. If omitted variables are uncorrelated with the independent variables included in the regression, the OLS regression will produce an unbiased estimate. However, when omitted variables are correlated with the independent variables included in the regression, OLS regression produces inconsistent and biased estimates, known as omitted variable bias (OVB).

Omitted Variable Bias occurs when the regressor is correlated with the omitted variable AND the omitted variable is a determinant of the dependent variable. OVB prevents the estimator from converging in probability to the true parameter value and the direction and the strength of the OVB are dependent on the correlation between the regressor and the error term.

**5:** A situation in which it would be reasonable to select a biased estimator in a statistical analysis is when we are interested in minimizing the MSE (mean squared error). There are a few ways to go about this, including ridge regression, the kNN classifier.

This makes it so the small bias is not even worth it since the MSE is larger but there are less variable classifiers. With this tradeoff, depending on the project, it may be more beneficial to select a more biased estimator. 

Unbiased estimators are often more variable whereas biased estimators are less variable. Hence even if I am not correct on average, for any instance of a biased estimator I would be close to the true value.
